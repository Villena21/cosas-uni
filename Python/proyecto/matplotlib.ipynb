### 0. Importar matplotlib

Comenzaremos importando [`matplotlib.pyplot`](https://matplotlib.org/stable/tutorials/introductory/pyplot.html).

Porque `pyplot` es un submódulo para crear gráficos interactivos de forma programática.

`pyplot` suele importarse con el alias `plt`.

### Dos formas de crear gráficos

Hay dos formas principales de crear gráficos en matplotlib.

1. `matplotlib.pyplot.plot()` - Recomendado para gráficos simples (por ejemplo, x e y).
2. `matplotlib.pyplot.XX` (donde XX puede ser uno de muchos métodos, esto se conoce como la API orientada a objetos) - Recomendado para gráficos más complejos (por ejemplo, [`plt.subplots()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html#matplotlib.pyplot.subplots) para crear múltiples gráficos en la misma figura; hablaremos de esto más adelante).

Ambos métodos suelen construirse partiendo de `import matplotlib.pyplot as plt` como base.

Empecemos con algo simple.

```python
plt.plot();
```

```python
plt.plot()
plt.show()
```

```python
plt.plot([1, 2, 3, 4]);
```

```python
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]
```

Algunas cosas rápidas sobre un gráfico:

* `x` es el eje horizontal.
* `y` es el eje vertical.
* En un punto de datos, `x` suele ir primero, por ejemplo, `(3, 4)` sería `(x=3, y=4)`.
* Lo mismo ocurre en [`matplotlib.pyplot.plot()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html), `x` va antes que `y`, por ejemplo, `plt.plot(x, y)`.

```python
plt.plot(x, y)
plt.show()
```

### 1. Primer ejemplo

Importamos la librería *NumPy* para crear datos

```python
import numpy as np
```

Ahora vamos a crear datos con *NumPy* que luego representaremos en una gráfica. La variable $x$ contiene un rango de valores entre $0$ y $2\pi$. La variable $y$ contiene el seno de $x$.   

```python
x = np.arange(0, 2*np.pi, 0.01)
y = np.sin(x)
```

Y por último representamos los datos anteriores en una gráfica. En la siguiente celda se realizan las siguientes tareas:

- Se crea la figura en la que se va a dibujar el gráfico con la llamada a <a href="http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.figure">plt.figure</a>. Se puede especificar el tamaño con el argumento *figsize*.

- Se crea el gráfico con la llamada a <a href="http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.plot">plt.plot</a>. Hay que especificar los valores *x* e *y*, que deberán ser arrays o similares de igual longitud, y opcionalmente una cadena que indica el estilo del gráfico. En el ejemplo se usa '-r', '-' para indicar que se quiere una línea continua y 'r' para indicar color rojo.

- Se dibuja una cuadrícula con <a href="http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.grid">plt.grid</a>, se crea un título para el gráfico con <a href="http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.title">plt.title</a> y se rotulan los ejes con <a href="http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.xlabel">plt.xlabel</a> y <a href="http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.ylabel">plt.ylabel</a>.


```python
# dibujamos la grafica

plt.figure(figsize=(8, 4))
plt.plot(x, y, "--r") #r: es para rojo, g:verde y la b:azul, --: es para que sea discontinua, -: continua
plt.grid(True)
plt.title('Grafica de la funcion seno')
plt.xlabel('X')
plt.ylabel('sen(x)')
plt.show()
```

### 2. Gráficos de barras

Para dibujar gráficos de barras utilizamos la función <a href="http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.bar">plt.bar</a>. Es necesario pasarle como argumentos los valores *x* e *y*, y opcionalmente etiquetas para las barras, el alineamiento de estas con respecto a los valores de *x*, el color de las barras, etc. 

A continuación se muestran algunos ejemplos.

```python
x = np.arange(8)
y = np.random.randint(1, 100, 8)

print(x)
print(y)
```

```python
plt.figure()
plt.bar(x, y, color = "g")
plt.title('Grafica de barras')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
plt.figure()
plt.bar(x, y, align='center', color = "purple", tick_label = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'])
plt.grid(True)
plt.title('Grafica de barras')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

Es posible también dibujar gráficos de barras horizontales utilizando la función <a href="http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.barh">plt.barh</a>.

```python
plt.figure()
plt.barh(x, y, align='center', color = "purple", tick_label = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'])
plt.grid(True)
plt.title('Grafica de barras')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

### 3. Histogramas

La función <a href="http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.hist">plt.hist</a> dibuja histogramas. Sus argumentos principales son un array o secuencia que contiene los datos, y el número de cajas o *bins* (por defecto igual a 10). La función divide el rango de valores de los datos en un conjunto de cajas equiespaciadas, cuenta el número de puntos que caen en cada caja y dibuja el resultado usando barras verticales. 

A continuación se muestra un ejemplo.

```python
x = np.random.randn(10000) #randn crea daros aleatorios con distribucion normal
x
```

```python
plt.figure()
plt.hist(x, bins= 50, color = "orange", edgecolor = "black")
plt.title('Histograma')
plt.xlabel('X')
plt.ylabel('Frecuencia')
plt.show()
```

### 4. Gráficos de dispersión

La función <a href="http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.scatter">plt.scatter</a> se utiliza para representar diagramas de dispersión de dos variables (*x* e *y*). Como argumentos opcionales se puede especificar el color de los puntos o el tipo y tamaño del marcador.

A continuación se muestran dos ejemplos.
 

```python
x = np.random.randn(2000) #randn crea daros aleatorios con distribucion normal
y = np.random.randn(2000)
```

```python
plt.figure()
plt.scatter(x, y, color = "red", edgecolor = "black")
plt.title('Gráfico de dispersión')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
plt.figure()
plt.scatter(x, y, c = "red", edgecolor = "black", marker = "d", s = 100)
plt.grid(True)
plt.title('Gráfico de dispersión')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

### 5. Varios conjuntos de datos en el mismo gráfico

Algunas de las funciones anteriores se pueden utilizar para representar varios conjuntos de datos de manera simultánea. A continuación tienes dos ejemplos, uno para *plot* y otro para *scatter*. En ambos se utiliza <a href="http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.legend">plt.legend</a> para añadir al gráfico una leyenda que identifica cada una de las series dibujadas.

```python
x = np.arange(0, 2*np.pi, 0.01)
y = np.sin(x)
z = np.cos(x)

#grafico
plt.figure(figsize=(8, 6))
plt.plot(x, y, "-r", x, z, "--g")
plt.title('Grafica de la funcion seno y coseno')
plt.xlabel('X')
plt.legend(['sen(x)', 'cos(x)'])
plt.show()
```

```python
#datos

x0 = np.random.randn(200) #randn crea daros aleatorios con distribucion normal
y0 = np.random.randn(200)
x1 = np.random.randn(200) + 2
y1 = np.random.randn(200) + 2


#grafico
plt.figure()
plt.scatter(x0, y0, c = "purple", edgecolor = "black", marker = "o", s = 30)
plt.scatter(x1, y1, c = "orange", edgecolor = "black", alpha = 0.5, marker = "o", s = 30)
plt.title('Gráfico de dispersión')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend(['Grupo 1', 'Grupo 2'])
plt.show()
```

### 6. Varios gráficos en la misma figura

Es posible dividir una figura en varios paneles, y dibujar un gráfico en cada uno de ellos. Para conseguirlo se utiliza la función <a href="http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.subplot">plt.subplot</a>, con tres argumentos que indican el número de filas, el número de columnas y la posición en la que se quiere dibujar el siguiente gráfico. La posición empieza en 1 y se cuenta por filas, de izquierda a derecha.

Por ejemplo la llamada *plt.subplot(2, 2, 3)* dividirá la figura en 4 subfiguras (2 filas y 2 columnas) y hará que el siguiente gráfico se dibuje en la subfigura correspondiente a la fila 2 y la columna 1.

El siguiente ejemplo crea una figura con dos subplots.


```python
plt.figure(figsize=(10, 6))

#en la sub figura de la izquierda dibujamos la grafica del seno y coseno
plt.subplot(1, 2, 1)
plt.plot(x, y, "-r", x, z, "--g")
plt.title('Grafica de la funcion seno y coseno')
plt.xlabel('X')
plt.legend(['sen(x)', 'cos(x)'])

#en la sub figura de la derecha dibujamos el grafico de dispersión
plt.subplot(1, 2, 2)
plt.scatter(x0, y0, c = "purple", edgecolor = "black", marker = "o", s = 30)
plt.scatter(x1, y1, c = "orange", edgecolor = "black", alpha = 0.5, marker = "o", s = 30)
plt.title('Gráfico de dispersión')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend(['Grupo 1', 'Grupo 2'])
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.exp(x)
y4 = np.log(x + 1)  # añadir 1 para evitar log(0)

#graficos

fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(10, 6))

axs[0, 0].plot(x, y1, "-r")
axs[0, 0].set_title('Seno')

axs[0, 1].plot(x, y2, "-g")
axs[0, 1].set_title('Coseno')

axs[1, 0].plot(x, y3, "-m")
axs[1, 0].set_title('Exponencial')

axs[1, 1].plot(x, y4, "-c")
axs[1, 1].set_title('Logaritmo')

plt.show()
```

### 8. Gráficos con Pandas

Aunque utilizar Matplotlib directamente para dibujar los datos de un *DataFrame* de *pandas* no resulta difícil, la librería *pandas* simplifica las cosas haciendo que gran parte de la funcionalidad anterior esté integrada en métodos de *DataFrame*. Las llamadas a estos métodos encapsulan llamadas a funciones de *matplotlib* para representar la información contenida en los datos de la manera más apropiada. Sin pretender que sea una muestra completa de las posibilidades disponibles, a continuación se muestran algunos ejemplos típicos. Echa un vistazo a <a href="http://pandas.pydata.org/pandas-docs/version/0.18.1/visualization.html">esta web</a> si quieres ver más opciones de visualización con *pandas*.

```python
import pandas as pd

serie = pd.Series(np.random.randn(1000)).cumsum()
serie.plot(figsize=(8, 6), grid=True, title='Serie temporal acumulada')
plt.show()
```

```python
df = pd.DataFrame(np.random.randn(1000, 3), columns=['1', '2', '3']).cumsum()
df.plot(figsize=(8, 6), grid=True, title='DataFrame acumulado')
plt.show()
```

```python
df = pd.DataFrame(np.random.randint(0, 100, (6, 4)), columns=['1', '2', '3', '4']).cumsum()
df
```

```python
df.plot(kind='bar', figsize=(8, 6), grid=True, title='Pacientes')
```
