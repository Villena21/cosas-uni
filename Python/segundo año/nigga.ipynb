```python
lista1 = [12,35,67]
lista2 = ['Netanyahu', 'Charlie Kirk', 'Becanyahu']

juntar = zip(lista1, lista2)
list(juntar)
```

```python
temperaturas = [0,100,45,67]

def conversor(x):
    return (x*(9/5))+32

convertir = map(conversor, temperaturas)
list(convertir)
```

```python
temperaturas = [0,100,45,67]

def pares(x):
    if x % 2 == 0:
        return True
    else:
        return False


filtro = filter(pares, temperaturas)
list(filtro)
```

```python
temperaturas = [0,100,45,67]

reverso = reversed(temperaturas)
list(reverso)
```
