## Ejercicio 2: Operadores aritméticos
Si un gen tiene **1500 pares de bases (bp)** y otro gen tiene **2300 bp**, calcula:
1. La longitud total combinada.
2. La diferencia en longitud.
3. El promedio de longitudes.

```python
gen1 = 1500
gen2= 2300

longitud_total = gen1 + gen2

diferencia = abs(gen1 - gen2)

promedio = (longitud_total) / 2

print(longitud_total)
print(diferencia)
print(promedio)
```

## Ejercicio 3: Condicional simple
Supón que tienes un valor de **GC% = 60**. Un genoma se considera **rico en GC** si su porcentaje es mayor a 50.

**Tarea:** Escribe un condicional que imprima si el genoma es rico en GC o no.

```python
gc = 60

if gc > 50:
    print("El genoma es rico en GC")

else:
    print("El genoma no es rico en GC")
```
