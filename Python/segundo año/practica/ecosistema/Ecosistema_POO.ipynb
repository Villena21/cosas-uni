## Parte 1

Crea las siguientes clases:

- `Planta`: con un atributo `nombre` y un método `crecer()` que imprima un mensaje como `"La planta [nombre] está creciendo."`
- `Animal`: con un atributo `nombre` y un método `comer()` que imprima `"El animal [nombre] está buscando comida."`
- `Ambiente`: con un atributo `nombre` y con listas de `animales` y `plantas`.
  - Un método `mostrar_ecosistema()` que imprima qué seres hay en el ambiente.
  - Los métodos para añadir plantas o animales al ambiente.

**Nota:** El ambiente debe “contener” animales y plantas.


````python
class Planta:
    def __init__(self, nombre):
        self.nombre = nombre

    def crecer(self):
        print(f"La planta {self.nombre} está creciendo.")


class Animal:
    def __init__(self, nombre):
        self.nombre = nombre

    def comer(self):
        print(f"El animal {self.nombre} está buscando comida.")


class Ambiente:
    def __init__(self, nombre):
        self.nombre = nombre
        self.animales = []
        self.plantas = []

    def agregar_animal(self, animal):
        self.animales.append(animal)

    def agregar_planta(self, planta):
        self.plantas.append(planta)

    def mostrar_ecosistema(self):
        print(f"=== Ecosistema: {self.nombre} ===")
        print("Animales:", ", ".join(a.nombre for a in self.animales))
        print("Plantas:", ", ".join(p.nombre for p in self.plantas))


# Ejemplo de uso
sabana = Ambiente("Sabana")
sabana.agregar_animal(Animal("León"))
sabana.agregar_animal(Animal("Cebra"))
sabana.agregar_planta(Planta("Hierba"))
sabana.agregar_planta(Planta("Arbusto"))
sabana.mostrar_ecosistema()
````

Agrega un método `alimentar_animales()` en `Ambiente` que:
- Tome un animal y una planta al azar (puedes usar `random.choice()`, tendrás que importar el módulo `random`).
- Haga que el animal “coma” la planta, mostrando algo como:  
  `"El animal León se comió la planta Hierba."`

**Nota**: ten en cuenta el caso que no hayan animales ni plantas en el ambiente


````python
import random

class Ambiente:
    def __init__(self, nombre):
        self.nombre = nombre
        self.animales = []
        self.plantas = []

    def agregar_animal(self, animal):
        self.animales.append(animal)

    def agregar_planta(self, planta):
        self.plantas.append(planta)

    def alimentar_animales(self, anmal_ran, plnt_ran):
        random.choice(self.animales) = anmal_ran
        random.choice(self.plantas) = plnt_ran

        print(f'El animal {anmal_ran}, se comio {plnt_ran}')

    def mostrar_ecosistema(self):
        print(f"=== Ecosistema: {self.nombre} ===")
        print("Animales:", ", ".join(a.nombre for a in self.animales))
        print("Plantas:", ", ".join(p.nombre for p in self.plantas))

sabana = Ambiente("Sabana")
sabana.agregar_animal(Animal("León"))
sabana.agregar_animal(Animal("Cebra"))
sabana.agregar_planta(Planta("Hierba"))
sabana.agregar_planta(Planta("Arbusto"))
sabana.alimentar_animales(Animal, Planta)
````
