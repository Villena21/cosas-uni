## Parte 1

Crea las siguientes clases:

- `Planta`: con un atributo `nombre` y un método `crecer()` que imprima un mensaje como `"La planta [nombre] está creciendo."`
- `Animal`: con un atributo `nombre` y un método `comer()` que imprima `"El animal [nombre] está buscando comida."`
- `Ambiente`: con un atributo `nombre` y con listas de `animales` y `plantas`.
  - Un método `mostrar_ecosistema()` que imprima qué seres hay en el ambiente.
  - Los métodos para añadir plantas o animales al ambiente.

**Nota:** El ambiente debe “contener” animales y plantas.


````python
class Planta:
    def __init__(self, nombre):
        self.nombre = nombre

    def crecer(self):
        print(f"La planta {self.nombre} está creciendo.")


class Animal:
    def __init__(self, nombre):
        self.nombre = nombre

    def comer(self):
        print(f"El animal {self.nombre} está buscando comida.")


class Ambiente:
    def __init__(self, nombre):
        self.nombre = nombre
        self.animales = []
        self.plantas = []

    def agregar_animal(self, animal):
        self.animales.append(animal)

    def agregar_planta(self, planta):
        self.plantas.append(planta)

    def mostrar_ecosistema(self):
        print(f"=== Ecosistema: {self.nombre} ===")
        print("Animales:", ", ".join(a.nombre for a in self.animales))
        print("Plantas:", ", ".join(p.nombre for p in self.plantas))


# Ejemplo de uso
sabana = Ambiente("Sabana")
sabana.agregar_animal(Animal("León"))
sabana.agregar_animal(Animal("Cebra"))
sabana.agregar_planta(Planta("Hierba"))
sabana.agregar_planta(Planta("Arbusto"))
sabana.mostrar_ecosistema()
````


## Parte 2 

Agrega un método `alimentar_animales()` en `Ambiente` que:
- Tome un animal y una planta al azar (puedes usar `random.choice()`, tendrás que importar el módulo `random`).
- Haga que el animal “coma” la planta, mostrando algo como:  
  `"El animal León se comió la planta Hierba."`

**Nota**: ten en cuenta el caso que no hayan animales ni plantas en el ambiente


````python
import random

class Ambiente:
    def __init__(self, nombre):
        self.nombre = nombre
        self.animales = []
        self.plantas = []

    def agregar_animal(self, animal):
        self.animales.append(animal)

    def agregar_planta(self, planta):
        self.plantas.append(planta)

    def mostrar_ecosistema(self):
        print(f"=== Ecosistema: {self.nombre} ===")
        print("Animales:", ", ".join(a.nombre for a in self.animales))
        print("Plantas:", ", ".join(p.nombre for p in self.plantas))

    def alimentar_animales(self):
        if not self.animales or not self.plantas: # comprueba si las listas estan vacias
            print("No hay suficientes seres vivos para la alimentación.")
            return

        animal = random.choice(self.animales)
        planta = random.choice(self.plantas)
        print(f"El animal {animal.nombre} se comió la planta {planta.nombre}.")


# Ejemplo de uso
sabana = Ambiente("Sabana")
sabana.agregar_animal(Animal("Cebra"))
sabana.agregar_animal(Animal("León"))
sabana.agregar_planta(Planta("Hierba"))
sabana.agregar_planta(Planta("Arbusto"))
sabana.mostrar_ecosistema()
sabana.alimentar_animales()
````


## Parte 3

Crea dos subclases de `Animal`:

- `Herbivoro`: redefine el método `comer()` para aceptar solo objetos de tipo `Planta`.
- `Carnivoro`: redefine el método `comer()` para aceptar solo objetos de tipo `Animal`.

Luego, modifica `alimentar_animales()` para:
- Si el animal es herbívoro → come plantas.
- Si es carnívoro → come otros animales (no a sí mismo).


````python

class Herbivoro(Animal):
    def comer(self, alimento):
        if isinstance(alimento, Planta):
            print(f"El herbívoro {self.nombre} se comió la planta {alimento.nombre}.")
        else:
            print(f"El herbívoro {self.nombre} no puede comer eso.")


class Carnivoro(Animal):
    def comer(self, alimento):
        if isinstance(alimento, Animal) and alimento != self:
            print(f"El carnívoro {self.nombre} se comió al animal {alimento.nombre}.")
        else:
            print(f"El carnívoro {self.nombre} no puede comer eso.")


class Ambiente:
    def __init__(self, nombre):
        self.nombre = nombre
        self.animales = []
        self.plantas = []

    def agregar_animal(self, animal):
        self.animales.append(animal)

    def agregar_planta(self, planta):
        self.plantas.append(planta)

    def mostrar_ecosistema(self):
        print(f"=== Ecosistema: {self.nombre} ===")
        print("Animales:", ", ".join(a.nombre for a in self.animales))
        print("Plantas:", ", ".join(p.nombre for p in self.plantas))

    def alimentar_animales(self):
        for animal in self.animales[:]:
            if isinstance(animal, Herbivoro) and self.plantas:
                planta = random.choice(self.plantas)
                animal.comer(planta)
                self.plantas.remove(planta)

            elif isinstance(animal, Carnivoro) and len(self.animales) > 1:
                posibles_presas = [a for a in self.animales if a != animal]
                presa = random.choice(posibles_presas)
                animal.comer(presa)
                self.animales.remove(presa)


# Ejemplo de uso
sabana = Ambiente("Sabana")
cebra = Herbivoro("Cebra")
leon = Carnivoro("León")
sabana.agregar_animal(cebra)
sabana.agregar_animal(leon)
sabana.agregar_planta(Planta("Hierba"))
sabana.agregar_planta(Planta("Arbusto"))

sabana.mostrar_ecosistema()
sabana.alimentar_animales()
````


## Parte 4

Agrega un método `simular_dia()` en `Ambiente` que:
1. Imprima el número de día que corresponde (si se llama por segunda vez que sea el día 2).
2. Llame a `mostrar_ecosistema()`.
3. Ejecute varias rondas de alimentación.
4. Muestre los resultados.

Ejemplo de salida esperada:
```
=== Día 1 ===
=== Ecosistema: Sabana ===
Animales: León, Cebra
Plantas: Hierba, Arbusto
El animal Cebra se comió la planta Hierba.
El animal León se comió al animal Cebra.
=== Fin del día 1. ===
```


````python

class Ambiente:
    dia = 0

    def __init__(self, nombre):
        self.nombre = nombre
        self.animales = []
        self.plantas = []

    def agregar_animal(self, animal):
        self.animales.append(animal)

    def agregar_planta(self, planta):
        self.plantas.append(planta)

    def mostrar_ecosistema(self):
        print(f"=== Ecosistema: {self.nombre} ===")
        print("Animales:", ", ".join(a.nombre for a in self.animales))
        print("Plantas:", ", ".join(p.nombre for p in self.plantas))

    def alimentar_animales(self):
        for animal in self.animales[:]:
            if isinstance(animal, Herbivoro) and self.plantas:
                planta = random.choice(self.plantas)
                animal.comer(planta)
                self.plantas.remove(planta)

            elif isinstance(animal, Carnivoro) and len(self.animales) > 1:
                posibles_presas = [a for a in self.animales if a != animal]
                presa = random.choice(posibles_presas)
                animal.comer(presa)
                self.animales.remove(presa)

    def simular_dia(self):
        Ambiente.dia += 1
        print(f"\n=== Día {Ambiente.dia} ===")
        self.mostrar_ecosistema()
        self.alimentar_animales()
        print("=== Fin del día. ===\n")


# Ejemplo de uso
sabana = Ambiente("Sabana")
sabana.agregar_animal(Herbivoro("Cebra"))
sabana.agregar_animal(Carnivoro("León"))
sabana.agregar_planta(Planta("Hierba"))
sabana.agregar_planta(Planta("Arbusto"))

sabana.simular_dia()
sabana.simular_dia()
````
