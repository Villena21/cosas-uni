## Ejercicio 1
Dada una lista de secuencias de ADN (`["ATCG", "GGG", "TTAAC", "A", "CCGGA"]`), escribe un programa que filtre solo aquellas con longitud mayor o igual a 3.

```python
secuencias = ["ATCG", "GGG", "TTAAC", "A", "CCGGA"]
secuancia_mas_3 = []

for i in secuencias:
    if len(i) >= 3:
        secuancia_mas_3.append(i)

secuancia_mas_3
```

## Ejercicio 3
Muestra cómo un set elimina duplicados al convertir una lista con elementos repetidos. 

```python
lista = ['ratio', 'glasses', 'conclusion', 'conclusion', 'conclusion', 'conclusion']
conjunto = {}

for i in lista:
    if i not in conjunto:
        conjunto.update(i)

conjunto
```

## Ejercicio 7
Dada una secuencia de ADN como string (`"ATCGATTGAGCTCTAGCG"`), construye un diccionario que cuente cuántas veces aparece cada nucleótido (A, T, C, G).

```python
cadena = "ATCGATTGAGCTCTAGCG"

num_a = cadena.count('A')
num_c = cadena.count('C')
num_g = cadena.count('G')
num_t = cadena.count('T')

nucleotidos = {
    'A': num_a,
    'C': num_c,
    'G': num_g,
    'T': num_t
}

nucleotidos
```
