```python
respuestas = ['a', 'b', 'c']
correctas = ['a', 'x', 'c']

respuestas_correctas = []

for i in respuestas:
    for j in correctas:
        if i in j:
            respuestas_correctas.append(i)


respuestas_correctas
```
