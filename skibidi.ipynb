```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def saludar(self):
        print(f'hola soy {self.nombre} y tengo {self.edad} anos')

p1 = Persona('EP', '67')
p1.saludar()
```

```python
class Coche:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo

    def cambiar_modelo(self, nuevo_modelo):
        self.modelo = nuevo_modelo

    def ensenar(self):
        print(f'Coche: {self.marca} {self.modelo}')

pi = Coche('fof', 'saas')
pi.cambiar_modelo('shoosh')
pi.ensenar()
```

```python
class Cuenta:
    def __init__(self, saldo_inicial = 0):
        self._saldo = saldo_inicial

    @property
    def saldo(self):
        return self.saldo

    def depositar(self, monto):
        if monto < 0:
            print('error: no puedes depositar dinero negativo')
        else:
            self._saldo += monto

    def consultar_saldo(self):
        return self._saldo

si = Cuenta()
si.depositar(-30)
si.consultar_saldo()
```

```python
class Animal:
    def hacer_sonido(self):
        pass

class Perro(Animal):
    def hacer_sonido(self):
        print('woof')


class Gato(Animal):
    def hacer_sonido(self):
        print('hfuadshfashfaj')

lista = [Gato(), Perro()]

for i in lista:
    i.hacer_sonido()
```
