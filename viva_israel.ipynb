```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def saludar(self):
        print(f"Hola, soy {self.nombre} y tengo {self.edad} años.")

nombre = Persona('nigga', 67)
nombre.saludar()
```

```python
class Coche:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo

    def cambiar_modelo(self, nuevo_modelo):
        self.modelo = nuevo_modelo

    def imprimir(self):
        print(f'Coche: {self.marca} {self.modelo}')

car = Coche('six', 'seven')
car.cambiar_modelo('PUTA')
car.imprimir()
```

```python
class Cuenta:
    def __init__(self, saldo_inicial=0):
        self._saldo = saldo_inicial

    @property

    def saldo(self):
        return self._saldo

    def depositar(self, monto):
        self._saldo += monto

    def consultar_saldo(self):
        return self._saldo

dinero = Cuenta()
dinero.depositar(100)
dinero.consultar_saldo()
```

```python
class Animal:
    def hacer_sonido(self):
        pass

class Perro(Animal):
    def hacer_sonido(self):
        print('Woof!')

class Gato(Animal):
    def hacer_sonido(self):
        print('Miaw!')

animales = [Perro(), Gato(), Perro(), Gato(), Gato(), Gato(), Gato()]

for i in animales:
    i.hacer_sonido()
```
