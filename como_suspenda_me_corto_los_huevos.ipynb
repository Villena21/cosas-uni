```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def saludar(self):
        print(f'Soy {self.nombre} y tengo {self.edad}')

p3 = Persona('Sergi', 18)
p3.saludar()
```

```python
class Coche:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo

    def cambiar_modelo(self, nuevo_modelo):
        self.modelo = nuevo_modelo

    def ensenar(self):
        print(f'{self.marca} {self.modelo}')

car = Coche('BMW', 'sixseven')
car.cambiar_modelo('profesora')
car.ensenar()
```

```python
class Cuenta:
    def __init__(self, saldo_inical = 0):
        self._saldo = saldo_inical

    @property
    def saldo(self):
        return self.saldo

    def depositar(self, monto):
        self._saldo += monto

    def consultar_saldo(self):
        return self._saldo

sixseven = Cuenta()
sixseven.depositar(1000)
sixseven.depositar(1000)
sixseven.consultar_saldo()
```
